```python
import requests

smiles = "CCO"

cardiotox = requests.post(
    "http://localhost:8101/smi",
    json={"smiles": smiles},
    timeout=300,
)
cardiotox.raise_for_status()

kinase = requests.post(
    "http://localhost:18019/smi",
    json={
        "smiles": smiles,
        "property": [""],
    },
    timeout=300,
)
kinase.raise_for_status()

print("CardioTox:", cardiotox.json())
print("KinasePred:", kinase.json())
```
